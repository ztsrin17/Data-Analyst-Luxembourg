```python
# Testing libraries setup

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a tiny "fake" transaction dataset
data = {
    'amount': [10.5, 1000.0, 25.0, 5000.0, 15.75],
    'is_fraud': [0, 1, 0, 1, 0]
}
df = pd.DataFrame(data)

# Print the data
print("Dataframe loaded successfully!")
display(df)

# Create a quick plot
sns.set_theme()
plt.figure(figsize=(6, 4))
sns.barplot(x='is_fraud', y='amount', data=df)
plt.title('Average Transaction Amount: Normal (0) vs Fraud (1)')
plt.show()
```
